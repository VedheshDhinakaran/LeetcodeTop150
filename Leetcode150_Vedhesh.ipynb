```python
# 169. Majority Element

class Solution:
    def majorityElement(self, nums: list[int]) -> int:
        n=len(nums)
        hash={}
        for i in nums:
            if i not in hash:
                hash[i]=1
            else:
                hash[i]+=1
        for i in hash:
            if hash[i]>n//2:
                return i
```
